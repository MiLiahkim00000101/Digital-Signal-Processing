# 1. Реализуем Вейвлет Морле

## Гойда

```python
N = 20000

t = np.linspace(-1/10, 1/10, N)

alpha = 0.02
w_0 = 2 * np.pi * 50


wavelet_morlet = np.exp(-t ** 2 / alpha ** 2) * np.exp(1j * w_0 * t)

plt.plot(t, wavelet_morlet)
```

```python
dt = t[1] - t[0]
freq_t = np.fft.fftfreq(N, dt)
bounds = (0, 300)

wavelet_morlet_freq = np.fft.fft(wavelet_morlet)

plt.plot(freq_t, np.abs(wavelet_morlet_freq)/N*2)
plt.xlim(bounds)
```
